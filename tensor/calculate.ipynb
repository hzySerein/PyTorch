```python
import torch
import numpy as np
```

# 四则运算 add,sub,mul,div

```python
tensor = torch.randint(0,10,size=(2,3))
print(tensor+10) #不修改原数据
print(tensor)
```

```python
tensor = torch.randint(0,10,size=(2,3))
print(tensor.add(10)) #不修改原数据
print(tensor)
```

```python
tensor = torch.randint(0,10,size=(2,3))
print(tensor.add_(10)) # 修改原数据
print(tensor)
```

```python
# neg
tensor = torch.randint(0,10,size=(2,3))
print(tensor.neg()) #或者neg_
print(tensor)
```

```python
# **
tensor = torch.randint(0,10,size=(2,3))
print(tensor**2)
print(tensor)
```

```python
# pow pow_
tensor = torch.randint(0,10,size=(2,3))
print(tensor.pow(2))
print(tensor)
print(tensor.pow_(2))
print(tensor)
```

```python
# sqrt sqrt_
tensor = torch.tensor([4,9,16,25])
print(tensor.sqrt())
print(tensor)

tensor = tensor.double()
print(tensor.double().sqrt_())
print(tensor)
```

```python
# exp exp_
tensor = torch.tensor([1.0,2,3,4])
print(tensor.exp())
print(tensor)

print(tensor.exp_())
print(tensor)
```

# 元素级乘法

```python
tensor = torch.tensor([[1,2,3],[4,5,6]])
print(tensor*tensor)
```

# 矩阵乘法

```python
# 二维
tensor1 = torch.tensor([[1,2,4],[3,4,6]])
tensor2 = torch.tensor([[1,2],[3,4],[5,6]])
print(tensor1 @ tensor2)
print(torch.matmul(tensor1,tensor2))
print(torch.mm(tensor1,tensor2))
```

```python
# 三维张量乘法
tensor1 = torch.tensor([[[1,2,4],[3,4,6]],[[1,2,4],[3,4,6]]])
tensor2 = torch.tensor([[[1,2],[3,4],[5,6]],[[1,2],[3,4],[5,6]]])
print(tensor1 @ tensor2)
```

```python
# sum()
tensor = torch.tensor([[1,2,3],[4,5,6]])
print(tensor.sum())

# sum(dim)
print(tensor.sum(dim=0))  # 对第0维求和(列)
print(tensor.sum(dim=1))  # 对第1维求和(行)
```
