```python
import torch
import numpy as np
```

# 创建张量

# 创建具体数张量

```python
tensor1 = torch.tensor(10.8)
print(tensor1)
print(tensor1.size())
print(tensor1.dtype)
print()

tensor2 = torch.tensor([1,2,3])
print(tensor2)
print(tensor2.size())
print(tensor2.dtype)
print()

tensor3 = torch.tensor([[1,2,3],[4,5,6]])
print(tensor3)
print(tensor3.size())
print(tensor3.dtype)

```

```python
tensor = torch.tensor(np.array([[1,2,3],[4,5,6]]))
print(tensor)
```

# 创建指定形状的张量

```python
tensor = torch.Tensor(3,2,4)
print(tensor)
print(tensor.size())
print(tensor.dtype)
```

```python
tensor = torch.Tensor([[1,2,3],[4,5,6]])
print(tensor)
print(tensor.size())
print(tensor.dtype)
```

```python
tensor = torch.Tensor(np.array([[1,2,3],[4,5,6]]))
print(tensor)
print(tensor.size())
print(tensor.dtype)
```

```python
tensor = torch.Tensor(10)
print(tensor)
print(tensor.size())
```

# 生成指定类型的张量

```python
tensor = torch.FloatTensor([[1,2,3],[4,5,6]])
print(tensor)
print(tensor.dtype)

tensor = torch.DoubleTensor([[1,2,3],[4,5,6]])
print(tensor)
print(tensor.dtype)
```

```python
tensor = torch.IntTensor([[1,2,3],[4,5,6]])
print(tensor)
print(tensor.dtype)
```

```python
tensor = torch.tensor([[1,2,3],[4,5,6]],dtype = torch.int64)
print(tensor)
print(tensor.dtype)

tensor2 = torch.LongTensor([[1,2,3],[4,5,6]])
print(tensor2)
print(tensor2.dtype)
```

```python
tensor  = torch.BoolTensor([[1,2,3],[4,5,6]])
print(tensor)
print(tensor.dtype)
```

# 创建指定区间的张量

```python
tensor1 = torch.arange(0,1.5,0.6)
print(tensor1)

tensor2 = torch.arange(10,1,-1)
print(tensor2)
```

```python
tensor = torch.linspace(0,4,4)
print(tensor)
```

```python
tensor = torch.logspace(1,3,3,base = 10)
print(tensor)
```

# 按数值填充张量

```python
tensor = torch.full([2,3],5)
print(tensor)
```

```python
tensor = torch.zeros([2,3],dtype = torch.int32)
print(tensor)
```

```python
tensor = torch.ones([2,3])
print(tensor)
```

```python
tensor = torch.empty([2,3])
print(tensor)
```

```python
tensor = torch.ones([2,3])
print(tensor)

tensor2 = torch.zeros_like(tensor)
print(tensor2)
```

# 随机张量创建

```python
# rand(size)：创建一个指定形状的随机张量，张量的元素取值范围是[0,1]
tensor  = torch.rand([2,3])
print(tensor)
```

```python
tensor = torch.randint(0,10,[2,3])
print(tensor)

tensor2 = torch.randint_like(tensor,0,10)
print(tensor2)
```

```python
# normal 正态分布
tensor = torch.randn([2,3])
print(tensor)
```

```python
tensor= torch.normal(mean = 0,std = 1,size = [2,3])
print(tensor)
```
