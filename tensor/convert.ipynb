# 类型转换

```python
tensor1 = torch.tensor([[1,2],[2,3],[3,4]])
print(tensor1.dtype)

tensor1 = tensor1.half()
print(tensor1.dtype)

tensor1 = tensor1.type(torch.float32)
print(tensor1.dtype)

tensor1 = tensor1.double()
print(tensor1.dtype)

tensor1 = tensor1.to(torch.complex64)
print(tensor1)
print(tensor1.dtype)

tensor1 = tensor1.to(torch.int32)
print(tensor1.dtype)


```

# tensor和ndarray转换

```python
# ndarray转tensor
array = np.array([1,2,3])
array[0] = 3 # 共享地址
tensor = torch.from_numpy(array)
print(tensor)
```

```python
# ndarray转tensor
array = np.array([1,2,3])
array[0] = 3 # 共享地址
tensor = torch.tensor(array)
print(tensor)
```

```python
# tensor转换成ndarray
tensor = torch.randint(0,10,[2,3])
print(tensor)
array = tensor.numpy()
tensor[:,1] = 2
print(array)
print(tensor)
```

```python
# 不共享内存
tensor = torch.randint(0,10,[2,3])
array = tensor.numpy().copy()
tensor[:,1] = 2
print(tensor)
print(array)
```

```python
array = np.array([1,2,3])
tensor = torch.from_numpy(array.copy())
array[2] = 5
print(array)
print(tensor)

```

```python
# tensor和标量转换
tensor = torch.tensor([10])
print(tensor)

print(tensor.item())
```
